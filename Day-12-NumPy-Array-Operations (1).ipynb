# Day 12 - NumPy Array Operations

## Objective
Create two NumPy arrays and perform addition, subtraction, multiplication, and division operations.

## Tools
- Python
- NumPy
- Jupyter Notebook

```python
import numpy as np
```

```python
array1 = np.array([10, 20, 30, 40, 50])

print("Array 1:", array1)
```

```python
array2 = np.array([2, 4, 5, 8, 10])

print("Array 2:", array2)
```

```python
print("Shape of Array 1:", array1.shape)
print("Shape of Array 2:", array2.shape)
```

```python
addition = array1 + array2

print("Addition:", addition)
```

```python
subtraction = array1 - array2

print("Subtraction:", subtraction)
```

```python
multiplication = array1 * array2

print("Multiplication:", multiplication)
```

```python
division = array1 / array2

print("Division:", division)
```

```python
print("Array 1:", array1)
print("Array 2:", array2)

print("\nAddition:", array1 + array2)
print("Subtraction:", array1 - array2)
print("Multiplication:", array1 * array2)
print("Division:", array1 / array2)
```

## Vectorization

Vectorization allows NumPy to perform mathematical operations on entire arrays without explicitly using Python for loops.

For example:

`array1 + array2`

performs addition element-by-element automatically.

```python
result = array1 + array2

print("Vectorized Addition:", result)
```

```python
result_loop = []

for i in range(len(array1)):
    result_loop.append(array1[i] + array2[i])

print("Using Python loop:", result_loop)
print("Using NumPy vectorization:", array1 + array2)
```

## Interview Questions

### 1. What is vectorization?

Vectorization is the process of performing operations on an entire NumPy array without explicitly using Python loops.

### 2. What happens when two NumPy arrays are added?

Corresponding elements of the two arrays are added together.

Example:

`[1, 2, 3] + [4, 5, 6] = [5, 7, 9]`

### 3. Why is NumPy faster than ordinary Python lists for numerical operations?

NumPy uses optimized numerical operations and vectorization, which reduces Python-level loop overhead and makes numerical computations more efficient.

### 4. What is the shape of a one-dimensional NumPy array containing 5 elements?

The shape is `(5,)`.

### 5. What happens if two NumPy arrays have incompatible shapes?

NumPy attempts broadcasting when possible. If their shapes cannot be broadcast together, it raises a `ValueError`.
